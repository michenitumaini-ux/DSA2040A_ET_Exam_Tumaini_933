# ETL Transform Phase

This notebook performs the **Transform** phase of the ETL process.  
It includes:
1. Data Cleaning  
2. Data Standardization  
3. Data Enrichment  
4. Structural Transformation  
5. Data Categorization  
6. Saving Transformed Data  
7. Summary of the Transformation


```python
# Import required libraries
import pandas as pd
import os

# Set working directory (update this path if necessary)
os.chdir(r"C:\Users\USER\Desktop\Data mining and warehousing\ET_Exam_Tumaini_933")

# Load validated dataset
validated_df = pd.read_csv("data/validated_data.csv")

# Preview data
validated_df.head()
```

## 1. Data Cleaning
In this step, we handle missing values, duplicates, and inconsistencies to ensure data quality.


```python
# Check missing values
print("Missing values before cleaning:\n", validated_df.isnull().sum())

# Fill missing numeric values with column means
validated_df = validated_df.fillna(validated_df.mean(numeric_only=True))

# Remove duplicates
validated_df = validated_df.drop_duplicates()

print("\nMissing values after cleaning:\n", validated_df.isnull().sum())
print("Data cleaning complete")
```

## 2. Data Standardization
Standardize column names, text formatting, and ensure consistent data units or formats.


```python
# Standardize column names
validated_df.columns = validated_df.columns.str.lower().str.replace(' ', '_')

# Example: Standardize city names (title case)
if 'borrower_city' in validated_df.columns:
    validated_df['borrower_city'] = validated_df['borrower_city'].str.title()

print("Standardization complete")
```

## 3. Data Enrichment
Enhance the dataset by adding new calculated or derived columns.


```python
# Example: Add a new derived column (amount in millions)
if 'amount_borrowed' in validated_df.columns:
    validated_df['amount_millions'] = validated_df['amount_borrowed'] / 1_000_000

print("Enrichment complete")
```

## 4. Structural Transformation
Reorganize or restructure the dataset (e.g., reorder columns or merge datasets).


```python
import pandas as pd
import os

os.chdir(r"C:\Users\USER\Desktop\Data mining and warehousing\ET_Exam_Tumaini_933")
validated_df = pd.read_csv("data/validated_data.csv")
```

```python
# Example: Reorder columns if they exist in the dataframe
expected_cols = ['borrower_city', 'borrower_zip_code', 'amount_borrowed', 'amount_millions']

# Select only columns that exist
available_cols = [col for col in expected_cols if col in validated_df.columns]

# Reorder dataframe based on available columns
validated_df = validated_df[available_cols]

print("Structural transformation complete")
validated_df.head()
```

## 5. Data Categorization
Create meaningful categories or bins to simplify data analysis.


```python
# Example: Categorize borrowers based on the amount borrowed
if 'amount_borrowed' in validated_df.columns:
    bins = [0, 1_000_000, 5_000_000, 10_000_000, validated_df['amount_borrowed'].max()]
    labels = ['Low', 'Medium', 'High', 'Very High']
    validated_df['borrow_category'] = pd.cut(validated_df['amount_borrowed'], bins=bins, labels=labels)

print("Categorization complete")
```

## 6. Save Transformed Data
Save the final cleaned and transformed data into the `transformed/` folder.


```python
# Save the transformed dataset
validated_df.to_csv("transformed/transformed_full.csv", index=False)

# Simulate incremental dataset (30% of data)
incremental_df = validated_df.sample(frac=0.3, random_state=42)
incremental_df.to_csv("transformed/transformed_incremental.csv", index=False)

print("Transformed datasets saved successfully in 'transformed/' folder")
```

Provide a quick overview of what was done and how the dataset looks after transformation.


```python
print("===== Transformation Summary =====")
print(f"Total records after cleaning: {len(validated_df)}")
print(f"Columns: {list(validated_df.columns)}")
print("\nMissing values after cleaning:\n", validated_df.isnull().sum())
print("\nSample records:")
display(validated_df.head())
print("\nTransformed data saved successfully!")
```
